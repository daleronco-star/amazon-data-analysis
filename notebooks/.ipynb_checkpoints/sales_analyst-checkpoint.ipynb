```python
# ==============================
# 1. Import Libraries
# ==============================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="viridis")

plt.rcParams["figure.figsize"] = (12,6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 12
```

```python
# ==============================
# 2. Load Dataset
# ==============================

df = pd.read_csv("C:/Users/MSI/Desktop/ecommerce-sales-data-analysis/data/amazon.csv")

df.head()
```

```python
# ==============================
# 3. Dataset Overview
# ==============================

print("Columns:")
print(df.columns)

df.info()
```

```python
# ==============================
# 4. Data Cleaning
# ==============================

df = df.drop_duplicates()

print(df.isnull().sum())
```

```python
# ==============================
# 5. Clean Price Columns
# ==============================

df['discounted_price'] = df['discounted_price'].astype(str).str.replace('₹','').str.replace(',','')
df['actual_price'] = df['actual_price'].astype(str).str.replace('₹','').str.replace(',','')

df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')
df['actual_price'] = pd.to_numeric(df['actual_price'], errors='coerce')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
```

```python
# ==============================
# 6. Rename Columns
# ==============================

df = df.rename(columns={
    "product_name": "Product",
    "category": "Category",
    "discounted_price": "Discounted Price",
    "actual_price": "Actual Price",
    "rating": "Rating",
    "rating_count": "Rating Count"
})
```

```python
# ==============================
# 7. Extract Main Category
# ==============================

df['Main Category'] = df['Category'].str.split('|').str[0]
```

```python
# ==============================
# 8. Top Products Analysis
# ==============================

top_products = df.groupby('Product')['Rating Count'].sum().sort_values(ascending=False)

top_10 = top_products.head(10)

short_names = [name[:40] + "..." for name in top_10.index]
```

# ==============================
9. Top Products Visualization
# ==============================

```python
plt.figure()

sns.barplot(
    x=top_10.values,
    y=short_names,
    hue=short_names,
    palette="viridis",
    legend=False
)

plt.title("Top 10 Most Rated Products")
plt.xlabel("Number of Ratings")
plt.ylabel("Product")

plt.tight_layout()

plt.show()
```

# ==============================
10. Price Distribution
# ==============================

```python
plt.figure()

sns.histplot(
    df['Discounted Price'],
    bins=30,
    color='royalblue'
)

plt.title("Distribution of Product Prices")
plt.xlabel("Price")

plt.tight_layout()

plt.show()
```

# ==============================
 11. Rating Distribution
# ==============================

```python
plt.figure()

sns.histplot(
    df['Rating'],
    bins=20,
    color='darkgreen'
)

plt.title("Rating Distribution")

plt.tight_layout()

plt.show()
```

# ==============================
 12. Price vs Rating
# ==============================

```python
plt.figure()

sns.scatterplot(
    x='Discounted Price',
    y='Rating',
    data=df,
    alpha=0.6
)

plt.title("Price vs Product Rating")

plt.tight_layout()

plt.show()
```

# ==============================
 13. Discount Analysis
# ==============================

```python
df['Discount Percentage'] = ((df['Actual Price'] - df['Discounted Price']) / df['Actual Price']) * 100

plt.figure()

sns.histplot(
    df['Discount Percentage'],
    bins=30,
    color='purple'
)

plt.title("Distribution of Amazon Discounts")
plt.xlabel("Discount Percentage")

plt.tight_layout()

plt.show()
```

# ==============================
 14. Product Category Analysis
# ==============================

```python
category_counts = df['Main Category'].value_counts().head(10)

plt.figure()

sns.barplot(
    x=category_counts.values,
    y=category_counts.index,
    hue=category_counts.index,
    palette="magma",
    legend=False
)

plt.title("Top 10 Product Categories")
plt.xlabel("Number of Products")
plt.ylabel("Category")

plt.tight_layout()

plt.show()
```

# ==============================
15. Price vs Discount
# ==============================

```python
plt.figure()

sns.scatterplot(
    x='Actual Price',
    y='Discount Percentage',
    data=df,
    alpha=0.6
)

plt.title("Product Price vs Discount Percentage")

plt.xlabel("Actual Price")
plt.ylabel("Discount %")

plt.tight_layout()

plt.show()
```

# ==============================
 16. Correlation Analysis
# ==============================

```python
plt.figure()

sns.heatmap(
    df[['Actual Price','Discounted Price','Rating','Rating Count','Discount Percentage']].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.tight_layout()

plt.show()
```

# ==============================
 17. Mini Dashboard
# ==============================

```python
fig, axes = plt.subplots(2, 2, figsize=(14,10))

# -----------------------------
# Top Categories
# -----------------------------

category_counts = df['Main Category'].value_counts().head(10)

sns.barplot(
    x=category_counts.values,
    y=category_counts.index,
    ax=axes[0,0],
    hue=category_counts.index,
    palette="magma",
    legend=False
)

axes[0,0].set_title("Top Product Categories")
axes[0,0].set_xlabel("Number of Products")
axes[0,0].set_ylabel("Category")


# -----------------------------
# Price Distribution
# -----------------------------

sns.histplot(
    df['Discounted Price'],
    bins=30,
    ax=axes[0,1],
    color="royalblue"
)

axes[0,1].set_title("Product Price Distribution")
axes[0,1].set_xlabel("Price")


# -----------------------------
# Rating Distribution
# -----------------------------

sns.histplot(
    df['Rating'],
    bins=20,
    ax=axes[1,0],
    color="darkgreen"
)

axes[1,0].set_title("Rating Distribution")
axes[1,0].set_xlabel("Rating")


# -----------------------------
# Price vs Discount
# -----------------------------

sns.scatterplot(
    x='Actual Price',
    y='Discount Percentage',
    data=df,
    ax=axes[1,1],
    alpha=0.6
)

axes[1,1].set_title("Price vs Discount")
axes[1,1].set_xlabel("Actual Price")
axes[1,1].set_ylabel("Discount %")


plt.suptitle("Amazon Products Data Analysis Dashboard", fontsize=16)

plt.tight_layout()

plt.show()
```

## Key Insights

• Some products dominate the total number of ratings, indicating high popularity.

• Products with medium prices tend to receive more ratings.

• Most products have ratings between 4 and 5.

• Higher-priced products tend to have larger discounts.

• Certain product categories dominate the Amazon catalog.
